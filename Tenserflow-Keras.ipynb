```python
def mutiple_model_outputs():
    inp = Input(shape=(11,))

    h1 = Dense(64, activation='relu', name='Hidden-layer-1')(inp)
    h2 = Dense(8, activation='relu', name='Hidden-layer-2')(h1)
    h3 = Dense(4, activation='relu', name='Hidden-layer-3')(h2)

    out1 = Dense(1, activation='sigmoid', name='Output-Layer-1')(h3)
    out2 = Dense(1, activation='relu', name='Output-Layer-2')(h3)

    model = Model(inputs=inp, outputs=[out1,out2], name='second-model-multiple-outputs')
    return model
```

```python
multiple_model = mutiple_model_outputs()
```
